## P05 — CoT · Número máximo de tubetes de anestésico para criança

```python
ID = "P05"

v1 = """Criança de 20 kg. Anestésico: lidocaína 2% com epinefrina 1:100.000, tubete de 1,8 mL. Dose máxima: 4,4 mg/kg. Quantos tubetes no máximo? Responda apenas o número."""

rodar(ID, "v1", v1)
```

```python
# Olhe o output da v1 e ajuste o texto do 'problema' abaixo para o que você viu de verdade.
# Resposta esperada: 20 x 4,4 = 88 mg; 1 tubete = 36 mg; 88/36 = 2,44 -> 2 tubetes.
final = """Calcule a quantidade máxima de tubetes de anestésico para uma criança, passo a passo.
Dados:
- Peso: 20 kg
- Lidocaína 2% = 20 mg por mL
- Tubete: 1,8 mL
- Dose máxima: 4,4 mg/kg
Passos:
1. Calcule a dose máxima total em mg.
2. Calcule quantos mg há em 1 tubete.
3. Divida e arredonde para BAIXO (nunca para cima).
Mostre cada conta. Na última linha escreva: "Resposta final: N tubetes".
Finalize com "Confira com o protocolo da clínica antes de usar"."""

rodar(ID, "final", final, temperature=0.0, max_tokens=300)
```

```python
ficha(
    ID,
    tecnica="Chain-of-Thought",
    objetivo="Obter o cálculo da dose máxima com passos conferíveis, em vez de um número solto.",
    problema="Pedir só o número impede conferir a conta: o modelo erra ou arredonda para cima sem mostrar como chegou.",
    risco="Erro numérico em dose de anestésico é risco clínico real para o paciente.",
    mitigacao="Dados (peso, concentração, dose máxima) vão no prompt, não ficam por conta da memória do modelo; passos explícitos, arredondar para baixo e aviso para conferir o protocolo.",
)
```

## P06 — CoT · Hipótese de pulpite reversível ou irreversível

```python
ID = "P06"

v1 = """Paciente com dor intensa ao frio que passa em poucos segundos depois de tirar o estímulo. Sem dor espontânea e sem dor à percussão. Qual o diagnóstico?"""

rodar(ID, "v1", v1)
```

```python
# Olhe o output da v1 e ajuste o texto do 'problema' abaixo para o que você viu de verdade.
# Resposta esperada: pulpite reversível.
final = """Você apoia um estudante de odontologia a raciocinar sobre um caso. Pense passo a passo.
Caso: dor intensa ao frio que passa em poucos segundos depois de tirar o estímulo; sem dor espontânea; sem dor à percussão.
Critérios:
- Pulpite reversível: dor provocada, curta, cessa ao remover o estímulo.
- Pulpite irreversível: dor espontânea, ou que persiste após remover o estímulo, ou noturna.
Passos:
1. Liste os achados do caso.
2. Compare cada achado com os critérios.
3. Conclua qual é a hipótese mais provável.
4. Diga que dado adicional falta (ex.: radiografia, teste de vitalidade).
Se faltarem dados para concluir, escreva "não sei". Termine com: "Hipótese para estudo; confirme com exame clínico e radiográfico." """

rodar(ID, "final", final, temperature=0.1, max_tokens=300)
```

```python
ficha(
    ID,
    tecnica="Chain-of-Thought",
    objetivo="Mostrar o raciocínio clínico (achados -> critérios -> hipótese) em vez de um diagnóstico seco.",
    problema="Pergunta direta ('qual o diagnóstico?') recebe uma resposta sem justificativa, que não dá para conferir nem usar para estudar.",
    risco="O texto pode ser lido como diagnóstico definitivo e a hipótese virar conduta sem exame.",
    mitigacao="Critérios no prompt, rótulo de 'hipótese para estudo', instrução 'não sei' se faltar dado e lembrete de exame clínico/radiográfico.",
)
```
