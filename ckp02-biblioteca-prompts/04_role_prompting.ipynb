# 04 · Role prompting — Odontologia
A **mesma pergunta** respondida por duas personas. P07: endodontista professor falando com estudante. P08: dentista acolhedora falando com paciente ansioso.

```python
# !pip install ollama      # rode uma vez se ainda não instalou
from helpers import rodar, ficha
```

```python
PERGUNTA = "Paciente com pulpite irreversível no dente 36. Explique o tratamento indicado e o que esperar."
```

## P07 — Role · Endodontista professor para estudante

```python
ID = "P07"

v1 = PERGUNTA

rodar(ID, "v1", v1)
```

```python
# Olhe o output da v1 e ajuste o texto do 'problema' abaixo para o que você viu de verdade.
system = """Você é um endodontista sênior e professor, explicando para um estudante do 3º ano de odontologia.
Use termos técnicos corretos (pulpectomia, instrumentação, irrigação, obturação).
Responda em até 6 linhas, em tópicos.
Você é um apoio de estudo: não prescreva medicamentos e lembre que a conduta depende do exame clínico e radiográfico."""

rodar(ID, "final", PERGUNTA, system=system, temperature=0.3, max_tokens=300)
```

```python
ficha(
    ID,
    tecnica="Role prompting",
    objetivo="Resposta técnica, no nível de um estudante, sobre o tratamento da pulpite irreversível.",
    problema="Sem persona, a resposta fica num meio-termo: nem técnica o bastante para o estudante, nem simples para o paciente.",
    risco="A persona de especialista dá ar de autoridade e a resposta pode ser tomada como conduta clínica ou prescrição.",
    mitigacao="System prompt define a persona como apoio de estudo, proíbe prescrever e exige citar exame clínico/radiográfico.",
)
```

## P08 — Role · Dentista acolhedora para paciente ansioso

```python
ID = "P08"

v1 = PERGUNTA

rodar(ID, "v1", v1)
```

```python
# Olhe o output da v1 e ajuste o texto do 'problema' abaixo para o que você viu de verdade.
system = """Você é uma cirurgiã-dentista acolhedora, conversando com um paciente leigo e ansioso.
Não use jargão; use uma analogia simples. Tranquilize sem prometer que não vai doer.
Responda em até 6 linhas, em tom de conversa.
Você é um apoio de comunicação: não prescreva medicamentos e diga que o dentista vai confirmar tudo na consulta."""

rodar(ID, "final", PERGUNTA, system=system, temperature=0.3, max_tokens=300)
```

```python
ficha(
    ID,
    tecnica="Role prompting",
    objetivo="Resposta simples e tranquilizadora, para paciente leigo e ansioso, à mesma pergunta do P07.",
    problema="Sem persona, a resposta usa termos técnicos e não considera o medo do paciente.",
    risco="Tom acolhedor pode levar o modelo a prometer ausência de dor ou a minimizar o problema.",
    mitigacao="System prompt proíbe prometer ausência de dor e prescrever, e manda dizer que o dentista confirma tudo na consulta.",
)
```
