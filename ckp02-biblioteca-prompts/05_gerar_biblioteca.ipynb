# 05 · Gerar a biblioteca
Lê o `resultados.json` (preenchido pelos notebooks 01–04) e monta o arquivo `BIBLIOTECA.md` com as 8 fichas. Rode **depois** dos outros quatro.

````python
import json
from pathlib import Path

dados = json.loads(Path("resultados.json").read_text(encoding="utf-8"))
ORDEM = ["P01", "P02", "P03", "P04", "P05", "P06", "P07", "P08"]
faltando = [p for p in ORDEM if p not in dados or "final" not in dados[p]]
print("Fichas faltando:", faltando or "nenhuma")
````

````python
def bloco_etapa(titulo, etapa):
    """Mostra system (se houver), prompt, output e parâmetros de uma etapa."""
    if not etapa:
        return ""
    texto = f"**{titulo}**\n\n"
    if etapa.get("system"):
        texto += f"System:\n```\n{etapa['system']}\n```\n\n"
    texto += f"Prompt:\n```\n{etapa['prompt']}\n```\n\nOutput real:\n```\n{etapa['output']}\n```\n\n"
    return texto


def ficha_md(id_prompt, d):
    p = d["final"]["params"]
    texto = f"## {id_prompt} — {d['tecnica']}\n\n"
    texto += f"**Objetivo:** {d['objetivo']}\n\n"
    texto += bloco_etapa("Versão inicial (v1)", d.get("v1"))
    texto += f"**Problema identificado na v1:** {d['problema']}\n\n"
    texto += bloco_etapa("Iteração intermediária (v2)", d.get("v2"))
    texto += bloco_etapa("Versão final", d["final"])
    texto += f"**Parâmetros:** modelo `{p['modelo']}`, temperature {p['temperature']}, max_tokens {p['max_tokens']}\n\n"
    texto += f"**Segurança — risco:** {d['risco']}\n\n"
    texto += f"**Segurança — mitigação na versão final:** {d['mitigacao']}\n\n---\n\n"
    return texto
````

````python
saida = "# Biblioteca de Prompts — Odontologia\n\n"
for id_prompt in ORDEM:
    if id_prompt in dados and "final" in dados[id_prompt]:
        saida += ficha_md(id_prompt, dados[id_prompt])

Path("BIBLIOTECA.md").write_text(saida, encoding="utf-8")
print(f"BIBLIOTECA.md gerado com {len(ORDEM) - len(faltando)} fichas")
````
